# 1. Problema e perguntas

## 1.1 Contexto e objetivo

**Problema e Relevância:**  
O mercado de jogos é extremamente competitivo. A desenvolvedora pode criar um jogo incrível e ele não ser jogado, ou ser jogado apenas anos depois de seu lançamento, pois foi lançado na época errada. Por isso desenvolvedoras e publicadoras precisam saber como está o mercado dos jogos para tomarem as decisões corretas para o desenvolvimento de seus jogos.

**Unidade de Observação:**  
A unidade de observação é o registro das vendas e metadados de cada jogo lançado individualmente para um console específico por exemplo, a versão de Grand Theft Auto V no PS3 é diferente da versão do Grand Theft Auto V no PS4 no dataset.

**Objetivo Analítico:**
O objetivo do trabalho é conduzir um processo de analises de vendas dos consoles (PS3, PS4 e PS5) da Playstation, identificando os jogos mais populares de cada console por região, o impacto do gêneros do jogos ao longo das gerações, e verificar se existe uma correlação entre a nota da crítica (Metacritic) e o volume de vendas entre as regiões

Para garantir a rastreabilidade e a qualidade dos dados dentro do ambiente de desenvolvimento, os dados são estruturados seguindo os princípios da arquitetura Medalhão.
1. **Camada Bronze:** Contém os dados brutos, carregados diretamente do CSV original sem nenhuma alteração.
2. **Camada Silver:** Onde será tratado os valores nulos, a correção de inconsistências, a normalização de gêneros e a formatação adequada de tipos de dados para análise.
3. **Camada Gold:** Camada final onde será gerado as agregações analíticas e visualizações específicas para responder a cada pergunta do projeto

**Uso Preditivo Futuro:**  
A análise das vendas e das preferência por determinados gêneros em consoles anteriores servem como base para planejar, e projetar o potencial de vendas e a aceitação do mercado para futuros lançamentos de jogos em novos consoles.

## 1.2 Perguntas ou hipóteses

- **H1:** Como as vendas dos 5 jogos mais populares de cada console (PS3, PS4 e PS5) se distribuem geograficamente entre as regiões (América do Norte, PAL, Japão e Outras)? Existe algum console com forte apelo regional específico?
- **H2:** Existe uma associação significativa entre a geração do console e a preferência de gêneros de jogos? Ou seja, o perfil de gêneros mais vendidos mudou do PS3 para o PS5, ou os mesmos gêneros continuam dominando o mercado de consoles da Sony?
- **H3:** Existe uma correlação entre a nota da crítica (Metacritic) e o volume de vendas globais? Essa relação se mantém consistente ou difere entre as gerações de consoles?

**Premissas:**
- Assume-se que os dados do dataset represente uma amostra do comportamento real do mercado para os consoles analisados.
- Assume-se que as notas de avaliação do Metacritic presentes no dataset refletem adequadamente a recepção crítica média de cada jogo no recorte temporal registrado do dataset original.

**Restrições e Recorte:**
- O estudo limita-se estritamente aos consoles PlayStation 3 (PS3), PlayStation 4 (PS4) e PlayStation 5 (PS5), ignorando outras plataformas, e os consoles anteriores a esses.
- A análise está restrita aos dados coletados da última data de atualização do dataset (Outubro de 2025).

**Limitações:**
- Não é possível rastrear o perfil individual de quem comprou os jogos (idade, gênero ou hábitos de consumo)
- Não dá pra saber com certeza se uma nota alta no metacrit é o motivo de um jogo ter mais vendas que outro, ou pelo jogo já ser de uma franquia famosa.
